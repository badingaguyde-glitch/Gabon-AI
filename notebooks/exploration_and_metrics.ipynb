# Gabon Domain LLM Dataset — Exploration & Metrik Analizi
Ce notebook permet d'explorer les données nettoyées du **Domain Corpus v1.0** et de l'**Instruction Dataset v1.0**.

```python
import json
from pathlib import Path
import pandas as pd

# 1. Chargement d'un extrait du corpus d'entraînement
train_path = Path('../data/final/train.jsonl')
docs = []
with open(train_path, 'r', encoding='utf-8') as f:
    for i, line in enumerate(f):
        if i >= 500:
            break
        docs.append(json.loads(line))

df_corpus = pd.DataFrame(docs)
print(f"Échantillon chargé : {len(df_corpus)} documents")
df_corpus[['document_id', 'title', 'subtopic', 'char_count', 'token_count']].head()
```

```python
# 2. Répartition des sous-thématiques
df_corpus['subtopic'].value_counts()
```

```python
# 3. Chargement de l'Instruction Dataset
inst_path = Path('../data/final/instruction_v1.0.jsonl')
instructions = []
with open(inst_path, 'r', encoding='utf-8') as f:
    for i, line in enumerate(f):
        instructions.append(json.loads(line))

df_inst = pd.DataFrame(instructions)
print(f"Instructions totales : {len(df_inst)}")
df_inst['task_type'].value_counts()
```
